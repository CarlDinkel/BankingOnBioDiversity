```python
import pandas as pd
from pathlib import Path
import re

folder = Path(r"C:\Users\Carl\Desktop\population density (01-20)")

files = sorted(folder.glob("uga_pd_*_1km_ASCII_XYZ*"))

dfs = []

for file in files:
    year = int(re.search(r"(\d{4})", file.name).group(1))

    df = pd.read_csv(file, sep=r"\s+|,|;", engine="python")

    df.columns = df.columns.str.strip()

    df = df.rename(columns={
        "X": "lon",
        "Y": "lat",
        "Z": f"density_{year}"
    })

    df = df[["lon", "lat", f"density_{year}"]]

    dfs.append(df)

    print(year, df.shape,
          "lon:", df["lon"].min(), df["lon"].max(),
          "lat:", df["lat"].min(), df["lat"].max())

wide = dfs[0]

for df in dfs[1:]:
    wide = wide.merge(df, on=["lon", "lat"], how="outer")

wide = wide.sort_values(["lat", "lon"]).reset_index(drop=True)

wide.to_csv(folder / "population_density_wide.csv", index=False)
wide.to_excel(folder / "population_density_wide.xlsx", index=False)

print("Final wide shape:", wide.shape)
```
