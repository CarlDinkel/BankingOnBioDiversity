```python
# CSB buffers
import geopandas as gpd
from rasterstats import zonal_stats
from pathlib import Path

buffer_path = r"C:\Users\Carl\Desktop\GIS\qgis layers\30km_circles.shp"
raster_folder = Path(r"C:\Users\Carl\Desktop\spam\2005_A")
out_csv = r"C:\Users\Carl\Desktop\GIS\spamoutput\spam2005_buffer_stats.csv"
out_gpkg = r"C:\Users\Carl\Desktop\GIS\spamoutput\spam2005_buffer_stats.gpkg"

buffers = gpd.read_file(buffer_path)

buffers = buffers[buffers.geometry.notna()].copy()

buffers["geometry"] = buffers.geometry.make_valid()

buffers = buffers[~buffers.geometry.is_empty].copy()

buffers = buffers.to_crs("EPSG:4326")

# test 3 crop rasters first
#rasters = list(raster_folder.glob("*_A.tif"))[:3]
rasters = list(raster_folder.glob("*_A.tif"))

for raster in rasters:
    crop = raster.stem.split("_")[-2].lower()
    print("Processing:", crop)

    stats = zonal_stats(
        buffers,
        str(raster),
        stats=["sum", "mean"],
        all_touched=True
    )

    buffers[f"{crop}_sum"] = [s["sum"] for s in stats]
    buffers[f"{crop}_mean"] = [s["mean"] for s in stats]

buffers.to_file(out_gpkg, driver="GPKG")
buffers.drop(columns="geometry").to_csv(out_csv, index=False)

print("Done")
```
